### Escape sequences in string

```python
# i):-// (backslash)
raw="C:\\users\\new_folder"
print(raw)
```

```python
# ii):- \' (single quote)
# hame ye printkana h:- It's my plasure
x="It's my Homework"
y='It\'s my Homework'
print(x)
print(y)
```

```python
# iii):- \" (double quote)
a='''It"s my Homework'''
print(a)
b='It\"s my homewrok'
print(b)

```

```python
#iv) \n  (new line)
c="Hello\nshivanshu"
print(c)
```

```python
#v) \t (tab)
d="Hello\tShivanshu"
print(d)
```

```python
# vi) \r carriage return
e='Thakur\rShivanshu'
print(e)
```

```python
# vii) \b (backspace)
f='Thakur\bShivanshu'
print(f)
```

```python
# viii) \uxxxx unicode 16-bit hex
g='\u2764'
print(g)

# ix) \uxxxxxxxx unicode 32-bit hex
# x) \xhh
```
