#### iii):-if-elif-else statements:-

```python
maths_mark=int(input("Enter the maths_marks"))
print("Entered the maths_marks is:-",maths_mark)
physics_marks=int(input("Enter the Physics_marks is:-"))
print("Entered the physics_marks is:-",physics_marks)
English_marks=int(input("Entered the English_marks is:-"))
print("Entered the English_marks is:-",English_marks)
chemistry_marks=int(input("Enter the chemistry_marks "))
print("Entered the chemistry_marks is:-",chemistry_marks)
Pd=int(input("Enter the Physicsal_Eduction_Marks"))
print("Entered the physical_education_marks is:-",Pd)
Total_Marks=maths_mark+physics_marks+English_marks+chemistry_marks+Pd
print("Total Marks is:-",Total_Marks)
Percentage=Total_Marks/5
print("Percentage is:-",Percentage)
if Percentage>=75 and Percentage<=100:
    print("First Division")
elif Percentage>=55 and Percentage<75:
    print("Second Division")
elif Percentage>=35 and Percentage<55:
    print("Third Division")
elif Percentage>=0 and Percentage<34:
    print("Are you fail try next time")
else:
    print("Enter valid marks")
```

#### Nested_Conditions:-

```python
x = 10
if x > 0:
    if x % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
```

#### Match case:-

```python
command=input("Enter the command:-").upper()
match command:
    case "START":
        print("Starting....")
    case "STOP":
        print("Stopping.....")
    case "QUIT" | "EXIT":
        print("Quitting....")
    case _:
        print("unkown command")
```
