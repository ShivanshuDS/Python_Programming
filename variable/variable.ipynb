#### Assigning the variable in many forms:-

##### 1.Basic Assignment of variable:-

```python
name="Shivanshu"
roll_no=34
salary=1550.50
```

##### 2.Multiple Assignment of variable:- it is also known as Tuple unpacking

```python
x,y,z= 12,14,15
print(x)
print(y)
print(z)
# iski asli power unpacking h
print()

cordinates=4,5
a,b=cordinates
print(a)
print(b)
```

```python
first,*rest,last=[10,20,30,40,50]
print(first)
print(rest)
print(last)
```

##### 3. Chained Assignment 

```python
x=y=z=12
print(x)
print(y)
print(z)
```

##### Swapping

```python
a,b=10,20
print(a,b)
a,b=b,a
print(a,b)
```
