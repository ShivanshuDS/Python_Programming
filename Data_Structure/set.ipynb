#### Creating the set:-

```python
# emptpy sets:-
st={}
print("empty set:-",st,type(st))# ye ese create nahi hoga

st=set()
print("empty set:-",st,type(st))

# numbers:-
st={21,34,12,21}
print("Set:-",st,type(st))

# mixed:-
st={34,"Shivanshu",67}
print("Mixed Set:-",st,type(st))
```

#### Set() constructor

```python
st=(12,34,56)
print(st,type(st))
s=set(st)
print(st,type(s))
print()

lst=['ram',23,45,23]
print(lst,type(lst))
l=set(lst)
print(l,type(l))
```

#### Set Comprehensions;-

#### Set Properties

#### Basic Operation of set
1.Union( | or union() )<br>
2.Intersection( & or intersection())<br>
3.Difference(- or difference())<br>
4.Symmetric Difference(^ or symmetric_difference())<br>
5.Subset and Superset(<= / issubset() and >= / issuperset)<br>
6.disjoint set(isdisjoint)

```python
a={12,13,45}
b={34,12,44}

# Union
print("Union of a | b:-",a|b)

#intersection
print("Intersection:-",a & b)

# difference
print("difference:-",a-b)

#subset
print("Subset:-",a<=b)

#superset
print("Superset:-",a>=b)

#disjoint set
print('disjoint set is:-',a.isdisjoint(b))
```

#### Set Methods:-
1.Adding Elements<br>
----Add<br>
----Update<br>
2.Removing Elements:-<br>
----remove<br>
----discard<br>
----pop<br>
----clear


##### 1.Adding Elements:-

```python
st={12,34,54,32}
print(st)
print()

 #Add
st.add("Shivanshu")
print("Add one element in the set is:-",st)

# update
st.update("Thakur")
print("Update the set",st)
```

##### 2.Removing Elements:-


```python
# remove:-
print(st)
print()
st.remove(12)
print("Remove the element:-",st)
print()

#pop
st.pop()
print("pop element from set by default:-",st)
# st.pop(3) # this part give error bcz set are unordered  
# print("pop the particular elements:-",st)
```

```python
#discard :- ye error nahi dega isse hmm particular element ko remove karte h
st.discard('h')
print("discard set :-",st)

#clear
st.clear()
print("clear the set:-",st)
```
