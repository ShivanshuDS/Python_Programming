## Tuple:-
1.Tuple is a immutable,ordered sequence of elements. it is hold the items and allows the duplictes.<br>
2.Tuple can be created Parenthesis '()'.<br>
3.Tule is faster than list bcz it is immutable.

```python
# Basic 
tup=(1,2,"HII")
print("tuple is:-",tup)
print(type(tup))
```

Creating the tuple:-

```python
empty=()
print("Empty Tuple is:-",empty)
print("type:-",type(empty))
print()

single_element_tup=(2,)# comma is required 
print("Single_Element_Tuple:-",single_element_tup)
print("type:-",type(single_element_tup))
print()

numbers=(12,14,45,32)
print("Tuple is:-",numbers)
print("type:-",type(numbers))
print()

mixed=(20,"Shivanshu","Pragati",20)
print("Tuple:-",mixed)
print("type:-",type(mixed))
print()

nested=((1,2),('yogesh','bhanu'))
print("nested tuple:-",nested)
print("type:-",type(nested))
```

```python
single=(1)
print(type(single)) # why 
```

#### tuple() constructor

```python
print(tuple("abc") ) 
print()      
print(tuple([1, 2, 3]))
print()     
print(tuple(range(3)))
print()
print(tuple({1, 2})) #order of arbitrary 

print()
print(tuple())            
```

#### tuple comprehensions

```python
tup=tuple(x**2 for x in range(5))
print(tup)
```

```python
tup=(1,3,4)*2
print(tup)
```

#### Tuple Properties:-

####  Packing and Unpacking of tuple:-

```python
# packing of tuple:-
tup=12,34,21
print("packing of tuple is:-",tup,":-","Type:-",type(tup))
print()

#unpacking of tuple:-
x=y=z=(10,20,30)
print("Unpacking of tuple is:-",x,":-","Type:-",type(x))
print("Unpacking of tuple is:-",y,":-","Type:-",type(y))
print("Unpacking of tuple is:-",z,":-","Type:-",type(z))
```

use * to capture remaining items to the list:-

```python
first,*middle,last=("Pragati",12,'yogesh','Shivanshu')
print(first,last,middle)
print()
print("First is:-",first)
print("Middle is-",middle)
print("last is:-",last)
```

#### Indexing and Slicing:-

```python
# indexing:-
tup=(12,'Shivanshu',34,'yogesh')
print(tup[0])
print(tup[3])
print(tup[-2])
print()

#Slicing
print(tup[1:4])
print(tup[1:-1])
```

#### Basic Operator in Tuple

```python
tup=(12,23,32,45,67)
print("Tuple:-",tup)
tup1=(90,64,45,65,43)
print("Tuple1:-",tup1)
```

 1.Concatination:-


```python
concatination=tup+tup1
print("Concatination:-",concatination)
```

2.Repeatation:-

```python
repeat=tup*2
print("Repeatation of tuple;-",repeat)
```

3.Membership:-<br>
-----in<br>
-----not in

```python
c=23 in tup
e=33 in tup
print("Membership (in) operator:-",c)
print("Membership (in) operator:-",e)
print()

d=36 not in tup
f=12 not in tup
print("Membership (not in) operator:-",d)
print("Membership (not in) operator:-",f)
```

4.len():- find the length of tuple:-

```python
print("Length of tuple1:-",len(tup1))

# 5.min()
print("Minimum number in tuple1:-",min(tup1))

# 6.max()
print("Maximum Number in tuple1:-",max(tup1))

#7.sum()
print("Sum of tuple1:-",sum(tup1))

#8.comparison(==)
print("comparison:-",tup==tup1)

#9.lexicographic(<,>)
print("Lexicographic:-",tup<tup1)
```
