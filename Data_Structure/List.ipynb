```python
lst=[13,4,6]
print("first list:-",lst)
lst1=[34,21,4]
print("second list:-",lst1)
print()

# concatination(+)
concate=(lst+lst1)
print("Concatination of two list:-",concate)
print()

#repetation
repeat=lst*2
print('repeat the first list:-',repeat)
print()
```

```python
#membership
print("memebership Operator in lst:-",13 in lst)
print("memebership Operator in lst:-",14 in lst1)
print()
print("memebership Operator in lst:-",20 not in lst1)
print("memebership Operator in lst:-",13  not in lst)
print()
```

```python
# any() Function:-it is return True if at least one element in the list is True.
numbers=[]
print(any(numbers))
print()

num=[1]
print(any(num))
print()

numbe=[0] # in this case me defination ke according "True" anna chhahiye but ye false de raha h  esa islye ki boolean data type me 0 means false hota h
print(any(numbe))
print()

numb=[0,12,34,23,54,"False"]
print(any(numb))
```

```python
# all() function:- it is return True only if every element in the list is True

number=[1,2,3,4]
print(all(number))
print()

num=[0,1,23,4]# ye false islye de raha h ki a/c to defination all elements are true in my list than answer will be true
print(all(num))
print()

numb=[-1,2,4,5,3]
print(all(numb))
```

Difference betwenn any() and all() function returns True when<br> any():-at least one element is True than answer will be True.<br>all():-Every Elements is True than answer will be true.

### List Methods:-
---
##### Adding Elemets:-
1.append(x)-add item to the end.<br>
2.insert(i,x):-add item in list in specifix  using index.<br>
3.extend(iterable):-add all items from iterable that means jinte item add krne ek saath value daal do.



```python
##Adding Elements:-

lst=["Anvi",134,"Anjali",452]
print("List:-",lst)

# using append(x)
lst.append("Pragati")
print("Update list using append:-",lst)
print()

# insert(i,x)
lst.insert(5,386)
print("update list using insert:-",lst)
print()

#extend()
lst.extend(["Shivanshu",321,"yogesh",43])
print("update list by using extend:-",lst)
```

##### Removing Elements:-
1.pop(i):-Removing the last in the lst.and give particular index than remove perticular index<br>
2.remove():-jo value remove kani h wo direct kar sakate h.<br>
3.clear():-all item from list can be remove.<br>
4.del:-remove the items throught the  index or slicing.


```python
# Removing elements:-

#1.pop(i):-
lst=['Anvi', 134, 'Anjali', 452, 'Pragati', 386, 'Shivanshu', 321, 'yogesh', 43]
print("list:-",lst)
lst.pop()
print()
print("Poped new lst:-",lst)
lst.pop(2)
print()
print("Poped list by using index:-",lst)
```

```python
#2.remove:-
lst=['Anvi', 134, 'Anjali', 452, 'Pragati', 386, 'Shivanshu', 321, 'yogesh', 43]
print("list:-",lst)
print()
lst.remove('yogesh')
print("Remove the item form list using remove();-",lst)
```

```python
#3.clear:-
print("list:-",lst)
print()
lst.clear()
print("using clear() new list:-",lst)
```

```python
#4. using del
lst=['Anvi', 134, 'Anjali', 452, 'Pragati', 386, 'Shivanshu', 321, 'yogesh', 43]
print("list:-",lst)
print()
del lst[5]
print("using del through index new list;-",lst)
print()
del lst[0:4]
print("using del through slicing new list:-",lst)
```

#### ordering and reversig:-
1.reverse()<br>
2.sort()<br>
3.sort(reverse=True)

```python
lst=[3,1,2,4]
print("list:-",lst)
print()
lst.reverse()
print("Reverse list:-",lst)
print()
lst.sort()
print("Sorted list:-",lst)
print()
lst.sort(reverse=True)#ye kiya karega pahle list ko sort karega fir reverse karega 
print("Sorted list with reverse is:-",lst)
```

#### key sorting-


```python
lst=['Bhanu','yogesh','anvi','anjali','kushi']
print("list:-",lst)
print()
lst.sort(key=len)
print("Sorted List according to length:-",lst)
```
