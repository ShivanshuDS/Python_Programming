#### 2.Relation/comparision operator:-
1.greater than(>)<br>
2.less than(<)<br>
3.greater than equal to (>=)<br>
4.less than equal to (<=)<br>
5.equal equal to (==)<br>
6.not equal to (!=)

```python
a=56
b=23
print("a greater than b:-",a>b)
print("a less than b:-",a<b)
print('a greater than equal to b:-',a>=b)
print('a less than equal to b:-',a<=b)
print('a equal equal to b:-',a==b)
print("a not equal to b:-",a!=b)
```

#### 3. Assignment Operator:-

```python
# Simple assignment:-
x=5 #that means x is assign the 5 value
x=y=z=4 # multiple assignment 
print(x)
print(y)
print(z)
print()
x,y=1,3 # tuple unpacking
print(x)
print(y)
```

```python
#argument assignment:-
x=8
print("Adding:-",x)
#adding
x+=1 #x=x+1
print(x)
x+=3 #x=x+3
print(x)
```

```python
x=10
print(x)
#subtraction
x-=5 # x=x-5
print('Subtraction:-',x)

x=20
#Multiplication
x*=2 #x=x*2
print('Multiplication:-',x)

x=13
#division
x/=3 #x=x*3
print("division:-",x)

x=13
#floor division
x//=3 #x=x//3
print("floor division:-",x)

x=10
#modulus 
x%=2
print("modulus is",x)

x=2
#exponential 
x**=2
print("Exponential:-",x)
```

#### 4.Logical Operator:-
1.and:-if both condition are true then answer will be true<br>
2.or:-if one condition are true then answer will be true<br>
3.not:-invert the true

```python
# and
x=30
y=20
c=x>y and y<x
print("x>y and y<x:-",c)
z=x>y and x<y
print("x>y and x<y:-",z)
```

```python
# or
x=39
print("value of x:-",x)
y=40
print("value of y:-",y)
c=x>y or x<y
print("x>y or x<y:-",c)
z=x>y or y<x
print("x>y or y<x:-",z)
```

```python
# not
x=True
print("the value of x",x)
print("the value of not x",not x)

```

#### 5.Bitwise Operator:-
1:-& (AND)<br>
2:-| (OR)<br>
3:-^ (XOR)<br>
4: ~ (BITWISE NOT)<br>
5:-<< (LEFT SHIFT)<br>
6:->> (RIGHT SHIFT)

```python
x=5
y=3
print("Bitwise AND:-",x&y)
print("Bitwise OR:-",x|y)
print("Bitwise XOR:-",x^y)
print("Bitwise Bitwise NOT:-",~y)
print("LEFT SHIFT:-",x<<y)
print("RIGHT SHIFT:-",x>>y)
```

#### 6.Identity Operator:-
1.is<br>
2.is not

```python
a=[12,23,45]
b=[45,27,38]
c=a
print("a is b:-",a is b)
print("a is c:-",a is c)
print('a is not b',a is not b)
```

#### 7.Membership Operator:-

```python
a=['Pragati','Shivanshu','Yogesh','Susmita']
print('Pragati' and  'Shivanshu' in a)
print('Ramu' in a)
print()

print("Ram" not in a)
print('Yogesh' and 'Susmita' not in a)
```
