### Basic Syntax:-

#### 1:- Print():-print() it is a built in function in python



```python
print("Hello,Shivanshu")#output:- Hello,Shivanshu
```

##### Full Syntax – print() function Parameters Explained ( full signature of print() function)

```python
print(*objects, sep=' ', end='\n', file=sys.stdout, flush=False)
```

#### (i):- *objects – Positional Arguments :- multiple iteams can be print in print() function

```python
print("Hello","World",123,[1,2,3])
print()

print("India",35,"Pakisthan",45)
print()

print("India","Pakisthan","Nepal","Srilanka")
```

#### (ii):-Sep – Separator:- sep parameter decide karta hai ki multiple objects ke beech mein kya aayega. Default ' ' (space) hai.

```python
print("India","Pakisthan","Nepal","Srilanka")# default 
print()

print("India","Pakisthan","Nepal","Srilanka",sep='/')
print()

print("India","Pakisthan","Nepal","Srilanka",sep='-')
```

#### (iii:-)end – Line End:-end parameter decide karta hai ki print ke baad line ke end mein kya character aayega. Default '\n' (newline) hota hai, isliye har print() nayi line mein jaata hai.

```python
print("Hello,")
print("Shivanshu")# esa kiyu hua ki doo print use kar rahe to next me jaa raha kiyu ki end=\n means nextline

# hum chahte ki ye mera ek hi line me aaye to kiya kare abbb hum tab end parameter use karo or modifiy kardo
print("Hello,",end=" ")
print("Shivanshu")
```
