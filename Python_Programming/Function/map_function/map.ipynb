## Map_Function:- Map function is a fucntion 
### syntax:-
--> map(fun,*iterable)<br>

----> *iterable me koi bhi list,tuple,dict,set etc.

```python
# suppose that you have a lst than double it using map function
L=[1,2,3,4,5,6,7]
map(lambda x:x*2,L)
```
