### Recursion Function:-function called itself that is called Recursion function

```python
# interviwer factorial without using the loop
# 5!=5,4,3,2,1
```

```python

# using recursion function 
def fact(n):
    if(n==1):
        return 1
    else:
        return n*fact(n-1)
       
fact(5)
```
