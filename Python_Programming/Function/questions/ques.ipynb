### Questions 1:- Wap to check whether a number is even is odd.

```python
number=int(input("Enter a Number:-"))
print("Entered Number:-",number)
def num(n):
    if n%2==0:
        print(f"{n} is even")
    else:
        print(f"{n} is odd")
num(number)
```

### Question2:- WAP to check whether a number is Positive or negative

```python
number=int(input("Enter a Number:-"))
print("Entered Number:-",number)
def check(n):
    if n>=0:
        print(f"{n} is Positive Number")
    else:
        print(f"{n} is negative Number")
check(number)
```

### Question 3:-WAP to Find the Largest among three Numbers.

```python
# using list so it is so simple
lst=[45,76,98]
print("Entered List:-",lst)
def largest(number):
    print(f"{max(number)} is Largest")
largest(lst)
    
```
