### Dictonary:-
1.A dictonary is a ordered,mutable collection that stores key value pairs.<br>
2.key must be unique or hashable.<br>
3.dictonary can be create curly braces { }.<br>

#### Creating the dictonary

```python
empty={}
print("Empty dict:-",empty,type(empty))
print()
name={'Name':['Aman','Raman','Shyam'],
      'Age':[20,19,18],
      'City':['Gurugram','Delhi','Nodia'] 
      }
print(name)
```

#### Dictonary Constructor:- dict()

```python
d=dict(name="Shivanshu",age=20)
print(d,type(d))

#maping 
s=dict({"a":1})
print(s)

#keyboard+iterable:-
s=dict({"a":1},b=2)
print(s)
```

#### Dictionary Comprehensions

```python
squares={x:x**2 for x in range(5)}
print(squares)
```

#### Note:- are you use  .append() in dictonary, to ye kiya karga tumhare all keys ki value same kar dega

```python
d = dict.fromkeys(["x","y"], [])
d["x"].append(1)
print(d)   # {'x': [1], 'y': [1]}   ← shared list
```

#### update():-

```python
details={'Name':['Pragati','Shivanshu','Yogesh','Susmita','Aditya'],
          'Age':[20,20,20,21,24],
          'Domain':['Data Analyst','Data Scientist','Full Stack','Business Analyst','DevOps developer'],}
print("full detailed dictonary:-",details)
print("Names key can access:-",details['Name'])
print("Age:-",details['Age'])
print("Domain:-",details['Domain'])
print()

# use update
details.update({'Course':['B.Tech','B.tech','BCA','M.Tech','MBA']})
print("Update dictonary is::-",details)
```

```python
d = {"a": 1}
print(d)
d.update({"b": 2, "a": 100})   
print(d)                               # {'a': 100, 'b': 2}
d.update([("c", 3)], d=4)       
print(d)                               # keyword args also 
```

#### Del statemets:-

```python
del d["a"]   # removes key 'a', raises KeyError if missing
print(d)
```

#### Dictonary views:-

```python
print(details)
print()
# all keys can be access
print(details.keys())
print()
# all values in dictonary can be access
print(details.values())
print()
# all items are show 
print(details.items())
```

### Note:- &(intersection),-(difference),|(union) it can be use in dictornary 

```python
d={"a":1,"b":2}
print(d)
c={'b':2,"C":4}
print(c)
k=d.keys() & c.keys()
print(k)
# f=d.values() & d.values() #ye error dega 
j=d.keys()-c.keys()
print(j)
```

#### Merging Dictonary:- by using |(union) this operator


```python
a={"x":1,"y":2}
b={"c":20,'z':3}
print(a)
print(b)
print()
c=a|b
print(c)
```
