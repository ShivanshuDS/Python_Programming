List comprehensions:-

```python
lst=[i**2 for i in range(5)]
print(lst)
print(type(lst))
```

```python
lst=[i for i in range(5)]
print(lst)
print(type(lst))
```

```python
lst=[i for i in range(10) if i%2==0]
print(lst)
print(type(lst))
```

```python
lst=[(x,y) for x in [1,2] for y in [3,4]]
print(lst)
print(type(lst))
```
