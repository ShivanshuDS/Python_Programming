## Loops:-
1.Loop is the backbone of automation programming.<br>
2.They allow you to execute a block of code repeatedly<br>
3.Python provides two primary  loop constructs:-<br>
----i):-while.<br>
---ii):-for.

##### i):- While:-
The while loop repeats a block of code as long as a given condition evaluates to True.

```python
count=0
while count<5:
    print(count,end=" ")
    count+=1
```

while-else:-<br>
--->The else block executes only if the loop terminates naturally (i.e., the condition becomes False). It does not execute if the loop is terminated by break.

```python
# Note:- I am use this so first of all execute while loop then execute the else bcz else can be use
# properly with if.
num = 5
while num > 0:
    print(num)
    num -= 1
else:
    print("Loop ended naturally!")  # Executes

```

```python
# With break:
# Note:- I use else with while usig break so else are skipped
num = 5
while num > 0:
    print(num)
    if num == 3:
        break
    num -= 1
else:
    print("This won't print because we broke out.")  # Skipped
```

##### ii):- For:-
The for loop in Python is fundamentally different from C/Java. It is an iterator-based loop designed to traverse any iterable object (list, tuple, string, dictionary, file, custom objects).

Syntax:-

```python
'''
for element in iterable:
    # Execute block for each element
    '''
```

```python
# List
for i in ["apple","mango","cherry"]:
    print(i,end='  ')
```

```python
lst=['Shivanshu','Aman','Bhanu','Shivam']
for i in lst :
    print(i)
```

```python
# print string character by character
for i in "Python":
    print(i)
```

```python
# tuple
tup=(1,13,45)
for i in tup:
    print(i,end='  ')
```

```python
# dictonary
dictonary={'Name':'Shivanshu','Age':20,'Course':'      B.Tech   ','Branch':'   CSE'}
for key in dictonary:
    print(key,end='       ')  # this is print only keys
print()

# To get value or items
for value in dictonary.values():
    print(value,end='  ') # this is print only value
print()

for i in dictonary.items():
    print(i)
```

for-else:-<br>
1.Identical logic to while-else: the else block executes if the loop completes without encountering a break.


```python
items = [1, 3, 5, 7]
for i in items:
    if i == 10:
        print("Found 10!")
        break
else:
    print("10 was not found in the list.")  # Executes bcz 10 is not mention in list so condition are false
                                                                        #then print else condition
```

#### Loop Control Statements(break, continue, pass)
Stataement<--------------------> Effect<-------------------------------------------> example<br>
break<-------> exit the innermost loop entirely.<----------------------------------> if error: break<br>
continue<----> skips the rest of the current iteration and jumps to next one.<---> if i % 2 == 0 : continue<br>
pass<--------> does absolutely nothing. A no-op placeholder.<-------------------> for x in range(5) : pass # TODO 

```python
# GOOD while with continue
i = 0
while i < 5:
    i += 1  # Increment happens BEFORE continue
    if i == 3:
        continue
    print(i)  # Output: 1, 2, 4, 5

```

##### Nested Loop:-
The inner loop completes all its iterations for every single iteration of the outer loop.

```python
print("i",end=' ')
print("j")
for i in range(5):
    for j in range(3):
        
        print(i,end=' ')
        print(j)
```

1.Strategy 1: Flag Variable

Strategy 2: for-else Trick (Elegant)
