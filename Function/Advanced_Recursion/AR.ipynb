```python
def mul(a,b):
    if b==1:
        return a
    else:
        return a+mul(a,b-1)
mul(5,6)
```

>Factorial Program:-->

```python

def fact(num):
    if num==1:
        return num
    else:
        return num*fact(num-1)
fact(4)
```

Check the palindrome in string using recursion:-

```python
def palin(text):
    if len(text)==1:
        print("Palindrome")
    else:
        if text[0]==text[-1]:
            palin(text[1:-1])
        else:
            print("not palin")
palin("madam")
palin("python")
palin("abba")# is line me code crash kar gaya h isko solve karne ke liye line no. 2 me change kar do let see next code.
```
