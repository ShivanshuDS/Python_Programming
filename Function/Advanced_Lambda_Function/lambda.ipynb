## Lambda Function:- 
1. It is a abnourmous function. it has no itself name.<br>
2. it is the one line function.
### Syntax:-
lambda input:expression<br>
lambda=keyword<br>

```python
# find square of a number using lambda
x=lambda x:x**2
print("square:-",x(2))

# find the sum of two numbers
sum=lambda x,y:x+y
print("Sum:-",sum(3,2))
```

```python

def return_sum():  
    lst=[11,14,23,56,78,45,29,28] 
    even_sum=0
    odd_sum=0
    div3_sum=0
    for i in lst:
        if i%2==0:
            even_sum=even_sum+i
        elif i%3==0:
                    div3_sum=div3_sum+i
        else:
            odd_sum=odd_sum+i
        
    return(even_sum,odd_sum,div3_sum)
return_sum()
```
