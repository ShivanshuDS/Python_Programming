## filter function:-
1.syntax:-<br>
list(filter(fun,*iterable))<br>
-->list():-iski jagh kuch bhi le sakte that means dict,list,tuple,set etc.

```python
# suppose that you have a list=[1,2,3,4,5,6,7] you find the grater number of 4.
l=[1,2,3,4,5,6,7]
filter(lambda x:x>4,l) # it convet list ya any
```

```python
list(filter(lambda x:x>4,l) ) # than convet it 
```

```python
fruits=["Apple","Orange","Mango","Guava"]
list(filter(lambda fruit:'e' in fruit,fruits))
```

### Differce B/w Map and Filter:-
map:-map function are work on all iteams of the list.
filter:- filter function are work on filtered/condition satisfy iteams on the list
