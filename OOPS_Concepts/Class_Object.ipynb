```python
class Arithmetic:
    def __init__(self,a,b):
        self.a=a
        self.b=b
        print("value of a:-",self.a)
        print("value of b:-",self.b)
    def add(self):
        print("addition:-",self.a+self.b)
    def sub(self):
        print("Subtraction:-",self.a-self.b)
    def mul(self):
        print("Multiplicatin:-",self.a*self.b)
obj=Arithmetic(30,20)
obj.add()
obj.sub()
obj.mul()
```

```python
import math
number=int(input())
class Shape:
    def __init__(self,n):
        self.n=n
        print("Given Number is:-",n)
    def square(self):
        print("Square is:-",self.n**2)
    def squareRoot(self):
        print("SquareRoot is:-",math.sqrt(self.n))
    def cube(self):
        print("Cube is:-",self.n**3)
obj=Shape(number)
obj.square()
obj.squareRoot()
obj.cube()
```

```python
# Write a code of atm machine
class Atm:
    def __init__(self):
        self.pin="" 
        self.balance=0
        self.menu()
    def menu(self):
        user_input=input('''What will you be perform
                            1.Enter 1  to create pin.
                            2.Enter 2 to deposit.
                            3.Enter 3 to withdraw.
                            4.Enter 4 to check_balance.
                            5.Enter 5 to exit.
                        ''')
        if user_input=="1":
            self.create_pin()
        elif user_input=="2":
            self.deposit()
        elif user_input=="3":
            self.withdraw()
        elif user_input=="4":
            self.check_balance()
        else:
            print("Exit")
    def create_pin(self):
        self.pin=input("Create Your pin:")
        print("Pin create successfully") 
    def deposit(self):
        created_pin=input("Enter your Pin:")
        if created_pin==self.pin:
            deposit_amount=int(input("Enter Your deposit_amount:"))
            self.balance=self.balance+deposit_amount
            print("deposit successfully")
        else:
            print("invalid pin")
    def withdraw(self):
        created_pin=input("Enter your pin:")
        if created_pin==self.pin:
            withdraw_amount=int(input("Enter Withdraw_amount:"))
            if withdraw_amount<=self.balance:
                self.balance=self.balance-withdraw_amount
                print("Withdraw successfully")
            else:
                print("insufficient balance")
        else:
            print("invalid pin")
    def check_balance(self):
        created_pin=input("Enter your pin:")
        if created_pin==self.pin:
            print(self.balance)
        else:
            print("invalid pin")

        
```
