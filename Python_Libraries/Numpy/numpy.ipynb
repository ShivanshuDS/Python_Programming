Create the zeroes in arrays

v=vertical,h=horizontal,

```python
arr1=np.array([10,20,30,40])
arr2=np.array([11,22,33,44])
print(arr1)
print(arr2)
```

```python
# combine the two array
# adding the two array in three ways
# 1.vstack:-multidimensional array can be create
arr3=np.vstack((arr1,arr2))
print(arr3)
np.shape(arr3)
```

```python
# 2.hstack:-create the singledimensional array
arr3=np.hstack((arr1,arr2))
print(arr3)
np.shape(arr3)
```

Numpy intersection and difference


```python
arr1=np.array([10,20,30,40])
arr2=np.array([11,22,33,44,10,40])
comman=np.intersect1d(arr1,arr2) # intersection means comman element 
comman
```

```python
difference=np.setdiff1d(arr1,arr2)
difference
# np.setdiff1d(arr2,arr1)
```

Numpy Array Mathematics

```python
n1=np.array([10,20,3])
n2=np.array([30,40,10])
res=sum([n1,n2])
res
```

```python
n1=np.array([10,20,25])
n2=np.array([30,40,10])
res=np.sum([n1,n2],axis=1)# jo axis=1 columns ke liye or axis=0 rows ke liye hota h
res
```
